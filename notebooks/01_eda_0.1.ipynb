**Obciążenia a churn**

Jak sytuacja życiowa klienta (senior, partner, dzieci/osoby na utrzymaniu) oraz koszty i staż wpływają na odejście? Dwa panele:

*Lewy*: odsetek odejść dla profilu gospodarstwa (kombinacja SeniorCitizen, Partner, Dependents). W etykiecie jest liczebność grupy.


*Prawy*: mapa ciepła odsetka odejść w zależności od stażu i opłaty miesięcznej.

```python
burden = df[["SeniorCitizen", "Partner", "Dependents",
             "MonthlyCharges", "tenure", "Churn"]].copy()


burden["profile"] = (
    burden["SeniorCitizen"].map({0: "", 1: "Senior, "})
    + "Partner: " + burden["Partner"]
    + ", Dzieci: " + burden["Dependents"]
)

profile_stats = (
    burden.groupby("profile")["Churn"]
    .agg(churn_rate="mean", n="size")
    .sort_values("churn_rate", ascending=False)
    .reset_index()
)
profile_stats["label"] = profile_stats["profile"] + " (n=" + profile_stats["n"].astype(str) + ")"


burden["tenure_bin"] = pd.cut(
    burden["tenure"], bins=[-1, 12, 24, 48, 72],
    labels=["0-12", "13-24", "25-48", "49-72"],
)
burden["charges_bin"] = pd.qcut(
    burden["MonthlyCharges"], q=4,
    labels=["najniższe", "niskie", "wysokie", "najwyższe"],
)

heat = burden.pivot_table(
    index="tenure_bin", columns="charges_bin",
    values="Churn", aggfunc="mean", observed=True,
)

fig, axes = plt.subplots(1, 2, figsize=(16, 6))

sns.barplot(data=profile_stats, y="label", x="churn_rate", color="steelblue", ax=axes[0])
axes[0].set_title("Odsetek odejść według profilu gospodarstwa")
axes[0].set_xlabel("Odsetek klientów, którzy odeszli")
axes[0].set_ylabel("")

sns.heatmap(heat, annot=True, fmt=".2f", cmap="Reds", vmin=0, vmax=1, ax=axes[1])
axes[1].set_title("Odsetek odejść: staż a opłata miesięczna")
axes[1].set_xlabel("Opłata miesięczna (kwartyle)")
axes[1].set_ylabel("Staż [miesiące]")

plt.tight_layout()
plt.show()
```

```python
burden = df[["SeniorCitizen", "Partner", "Dependents", "Contract"]].copy()

burden["profile"] = (
    burden["SeniorCitizen"].map({0: "", 1: "Senior, "})
    + "Partner: "
    + burden["Partner"]
    + ", Dzieci: "
    + burden["Dependents"]
)

profile_stats = (
    burden.groupby("profile")
    .size()
    .reset_index(name="n")
    .sort_values("n", ascending=False)
)

profile_contract = (
    pd.crosstab(burden["profile"], burden["Contract"], normalize="index") * 100
)
profile_contract = profile_contract.loc[profile_stats["profile"]]

plt.figure(figsize=(12, 6))
profile_contract.plot(
    kind="bar",
    stacked=True,
    colormap="Blues",
    edgecolor="black",
    linewidth=0.5,
)

plt.title(
    "Struktura typów umów w podziale na profile gospodarstwa domowego",
    fontsize=14,
    pad=15,
)
plt.xlabel("Profil gospodarstwa", fontsize=12)
plt.ylabel("Udział procentowy (%)", fontsize=12)
plt.legend(title="Typ umowy", bbox_to_anchor=(1.05, 1), loc="upper left")
plt.xticks(rotation=45, ha="right")
plt.ylim(0, 100)
plt.grid(axis="y", linestyle="--", alpha=0.5)

plt.tight_layout()
plt.show()
```

Najwyższy odsetek odejść notują zazwyczaj osoby samotne, młodsze (niebędące seniorami), bez osób na utrzymaniu, podczas gdy klienci mieszkający z rodzinami/partnerami i dziećmi są znacznie bardziej stabilni (wykazują niższy churn). Natomiast należy pamiętać, że w danych nie ma informacji ile churnu dotyczącego seniorów związane jest ze zgonami (tu zwracając uwagę w ilu grupach pojawiają się seniorzy - odsetek churnu w tej grupie jest znaczący). 

Nalezy zwrócić uwagę, że we wskazanych gospodarstwach dominują umowy krótkoterminowe, co wzmacja ryzyko.

#### **Odsetek odejść a liczba posiadanych produktów**

Czy im więcej produktów i wyższe koszta, tym wyższa szansa na churn?

```python

addons = ["OnlineSecurity", "OnlineBackup", "DeviceProtection",
          "TechSupport", "StreamingTV", "StreamingMovies"]

df["n_products"] = (
    df["PhoneService"].eq("Yes").astype(int)
    + df["InternetService"].ne("No").astype(int)
    + df["MultipleLines"].eq("Yes").astype(int)
    + sum(df[c].eq("Yes").astype(int) for c in addons)
)

by_count = (
    df.groupby("n_products")["Churn"]
    .agg(churn_rate="mean", n="size")
    .reset_index()
)

fig, ax = plt.subplots(figsize=(9, 5))

sns.barplot(data=by_count, x="n_products", y="churn_rate",
            color="steelblue", ax=ax)
for i, row in by_count.iterrows():
    ax.text(i, row["churn_rate"] + 0.01, f"n={row['n']}", ha="center", fontsize=8)

ax.set_title("Odsetek odejść według liczby produktów")
ax.set_xlabel("Liczba produktów klienta")
ax.set_ylabel("Odsetek klientów, którzy odeszli")
ax.set_ylim(0, by_count["churn_rate"].max() + 0.1)

plt.tight_layout()
plt.show()
```

Największy odestek występuje u klientów posiadających umiarkowaną liczbę usług (najwięcej przy 2, 3 i 4 usługach), najmniej osoby z 9 produktami. nistkie zagrożenie to również osoby z 1 usluga oraz 8 usługami. Hipoteza - im więcej produktów, tym korzystniejsze pakiety i większe przywiązanie. Dlaczego odsetek churnu najwyższy jest dla średnich grup? To potencjalni klienci do rozszeżenia, więc dlaczego jest taki odsetek odejść?

```python
def count_services(row):
  services = [
      'PhoneService',
      'MultipleLines',
      'InternetService',
      'OnlineSecurity',
      'OnlineBackup',
      'DeviceProtection',
      'TechSupport',
      'StreamingTV',
      'StreamingMovies',
  ]
  count = 0
  for s in services:
    val = row[s]
    if s == 'InternetService':
      if val in ['DSL', 'Fiber optic']:
        count += 1
    elif val == 'Yes':
      count += 1
  return count


df['ServiceCount'] = df.apply(count_services, axis=1)


df['Charge_Bin'] = pd.qcut(
    df['MonthlyCharges'],
    q=5,
    labels=['Bardzo niskie', 'Niskie', 'Średnie', 'Wysokie', 'Bardzo wysokie'],
)


pivot_table = pd.pivot_table(
    df,
    values='Churn',
    index='Charge_Bin',
    columns='ServiceCount',
    aggfunc=lambda x: (x == 'Yes').mean(),
)


plt.figure(figsize=(12, 6))
sns.heatmap(
    pivot_table,
    annot=True,
    fmt='.1%',
    cmap='Reds',
    cbar_kws={'label': 'Odsetek rezygnacji'},
)

plt.title(
    'Mapa ciepła: Odsetek rezygnacji w zależności od liczby usług i przedziału'
    ' kosztów',
    fontsize=13,
    pad=15,
)
plt.xlabel('Liczba posiadanych usług', fontsize=12)
plt.ylabel('Przedział opłat miesięcznych', fontsize=12)

plt.tight_layout()
plt.savefig('services_cost_heatmap.png', dpi=300)
plt.show()
```

Ekstremalnie wysoki churn (sięgający od $51.9\%$ do aż $70.7\%$) koncentruje się u klientów, którzy posiadają niewielką liczbę usług (2 do 3 produktów), a jednocześnie generują średnie lub wysokie opłaty miesięczne.

Mapa wyraźnie pokazuje, że wysoka cena sama w sobie nie jest głównym powodem odejścia – kluczowy jest stosunek kosztu do liczby otrzymywanych usług. Klienci płacący wysokie kwoty przy dużej liczbie usług (7–9 produktów) wykazują bardzo niski wskaźnik churnu (zaledwie od $1.2\%$ do $17.4\%$).

Hipoteza - klienci z grupy średniej są najbardziej narażeni na churn, szczególnie ci, którzy placą najwyższe koszta, ponieważ posiadają produkt, który stanowi główny składnik tych kosztów i jest powodem rezygnacji (to na bazie tego, że klienci posiadajacy 1 produkt płacą stosunkowo mało, ale już od 2 produktów koszty mogą mocno skoczyc w górę, podobnie churn). Ergo - istnieje produkt, który powoduje oddejścia klientów, potencjalnie jest z zakresu usług internetowych.

**Churn a uslugi internetowe**

```python
import matplotlib.ticker as ticker

df['Churn_num'] = (df['Churn'] == 'Yes').astype(int)


def count_services(row):
  services = [
      'PhoneService',
      'MultipleLines',
      'InternetService',
      'OnlineSecurity',
      'OnlineBackup',
      'DeviceProtection',
      'TechSupport',
      'StreamingTV',
      'StreamingMovies',
  ]
  count = 0
  for s in services:
    val = row[s]
    if s == 'InternetService':
      if val in ['DSL', 'Fiber optic']:
        count += 1
    elif val == 'Yes':
      count += 1
  return count


df['n_products'] = df.apply(count_services, axis=1)

addons = [
    'OnlineSecurity',
    'OnlineBackup',
    'DeviceProtection',
    'TechSupport',
    'StreamingTV',
    'StreamingMovies',
]

net_stats = (
    df.groupby('InternetService')['Churn_num']
    .agg(churn_rate='mean', n='size')
    .sort_values('churn_rate', ascending=False)
    .reset_index()
)

by_internet = (
    df[df['InternetService'] != 'Nie']
    .groupby(['InternetService', 'n_products'])['Churn_num']
    .agg(churn_rate='mean', n='size')
    .reset_index()
)
by_internet = by_internet[by_internet['n'] >= 30]

all_rows = pd.Series(True, index=df.index)
has_phone = df['PhoneService'] == 'Yes'
has_net = df['InternetService'] != 'No'

specs = [
    ('PhoneService', df['PhoneService'].eq('Yes'), all_rows),
    ('MultipleLines', df['MultipleLines'].eq('Yes'), has_phone),
    ('Internet DSL', df['InternetService'].eq('DSL'), all_rows),
    ('Internet Fiber', df['InternetService'].eq('Fiber optic'), all_rows),
] + [(c, df[c].eq('Yes'), has_net) for c in addons]

rows = []
for name, has, pop in specs:
  rows.append({
      'produkt': name,
      'grupa': 'ma',
      'churn_rate': df.loc[pop & has, 'Churn_num'].mean(),
      'n': (pop & has).sum(),
  })
  rows.append({
      'produkt': name,
      'grupa': 'nie ma',
      'churn_rate': df.loc[pop & ~has, 'Churn_num'].mean(),
      'n': (pop & ~has).sum(),
  })
product_effect = pd.DataFrame(rows)

product_order = (
    product_effect.groupby('produkt')['churn_rate']
    .max()
    .sort_values(ascending=True)
    .index
)

fig, axes = plt.subplots(3, 1, figsize=(12, 16))

sns.barplot(
    data=net_stats,
    x='InternetService',
    y='churn_rate',
    hue='InternetService',
    palette='Blues_r',
    ax=axes[0],
    legend=False,
)
for i, row in net_stats.iterrows():
  axes[0].text(
      i,
      row['churn_rate'] + 0.02,
      f"n={row['n']:,}\n({row['churn_rate']:.1%})",
      ha='center',
      fontsize=10,
      fontweight='bold',
  )
axes[0].set_title(
    '1. Odejścia według typu usługi internetowej',
    fontsize=13,
    pad=10,
)
axes[0].set_xlabel('Usługa internetowa', fontsize=11)
axes[0].set_ylabel('Wskaźnik rezygnacji', fontsize=11)
axes[0].yaxis.set_major_formatter(ticker.PercentFormatter(1.0))
axes[0].set_ylim(0, net_stats['churn_rate'].max() + 0.15)


sns.lineplot(
    data=by_internet,
    x='n_products',
    y='churn_rate',
    hue='InternetService',
    marker='o',
    linewidth=2.5,
    ax=axes[1],
)
axes[1].set_title(
    '2. Wpływ liczby produktów na churn w podziale na DSL i Fiber optic',
    fontsize=13,
    pad=10,
)
axes[1].set_xlabel('Liczba produktów klienta', fontsize=11)
axes[1].set_ylabel('Wskaźnik rezygnacji', fontsize=11)
axes[1].yaxis.set_major_formatter(ticker.PercentFormatter(1.0))
axes[1].grid(True, linestyle='--', alpha=0.5)
axes[1].legend(title='Rodzaj internetu')


sns.barplot(
    data=product_effect,
    y='produkt',
    x='churn_rate',
    hue='grupa',
    order=product_order,
    palette={'ma': '#2b5c8f', 'nie ma': '#bdc3c7'},
    ax=axes[2],
)
axes[2].set_title(
    '3. Wpływ poszczególnych produktów i dodatków na odsetek odejść (Ma vs Nie'
    ' ma)',
    fontsize=13,
    pad=10,
)
axes[2].set_xlabel('Wskaźnik rezygnacji', fontsize=11)
axes[2].set_ylabel('Produkt / Usługa', fontsize=11)
axes[2].xaxis.set_major_formatter(ticker.PercentFormatter(1.0))
axes[2].legend(title='Status klienta')

plt.tight_layout()
plt.show()
```

Hipoteza, że newralgiczny produkt znajduje się w grupie produktów internetowych znalazla potwierdzenie. Należy skupić się na *Internet Fiber/Fiber Optic*, gdzie widoczny jest churn w wysokości $41,92%$. W grupie klientów z 2, 3 i 4 produktami, czyli grupy o najwyższym churnie w stosunku do ilości produktów, pojawia się już jako 2 produkt. Jednocześnie należy zauważyć, że stnieje drastyczna przepaść w wskaźnikach rezygnacji w zależności od wybranej technologii dostępu do internetu. Klienci korzystający z Fiber optic rezygnują średnio w $41.9\%$ przypadków, klienci z internetem DSL mają churn na poziomie $19.0\%$ ($n=2,421$), a osoby bez internetu zaledwie $7.4\%$. W obu przypadkach widać natomiast duży spadek odejść klientów im więcej posiadają produktów - wygrywa tu pakietowość. Jednocześnie należy zauważy jak niski w stosunku do innych produktów jest churn dla *TechSupport* oraz *OnlineSecurity*. W związku z tym można założyć, że klienci posiadający je, szczególnie w kombinacji z innymi produktami internetowymi, chwalą sobie pomoc ze strony firmy w tym zakresie.

Wniosek - *Internet Fiber/Fiber Optic* jest głównym problemem dla firmy. Mimo że jest to najnowocześniejsza usługa, generuje dwukrotnie wyższy odpływ klientów niż DSL. Sugeruje to silną presję konkurencyjną na rynku światłowodowym, zbyt wysokie ceny lub niezadowolenie z jakości obsługi posprzedażowej w tym segmencie. 


**Sposób płatności a churn**
