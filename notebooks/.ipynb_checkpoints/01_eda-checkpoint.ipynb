## Czyszczenie  danych

1. W TotalCharges puste miejsca stają się NaN
2. Te 11 braków to klienci z tenure = 0, czyli łącznie zapłacili 0.
3. customerID to numer, pomaga przy sortowaniu danych
4. Target: Yes/No = 1/0 (1 = odszedł)

```python
df["TotalCharges"] = pd.to_numeric(df["TotalCharges"], errors="coerce")

df["TotalCharges"] = df["TotalCharges"].fillna(0)

df = df.drop(columns=["customerID"])

df["Churn"] = df["Churn"].map({"Yes": 1, "No": 0})
```

## Oglądanie danych

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
```

**Umowa a churn**
Czy churn zależy od typu umowy?

```python
fig, ax = plt.subplots(figsize=(7, 4))

sns.barplot(
    data=df,
    x="Contract",
    y="Churn",
    order=["Month-to-month", "One year", "Two year"],
    errorbar=None,
    ax=ax,
)

ax.set_title("Odsetek odejść według typu umowy")
ax.set_xlabel("Typ umowy")
ax.set_ylabel("Odsetek klientów, którzy odeszli")

plt.tight_layout()
plt.show()
```

Widać korelację - umowa 'Month-to-month' wykazuje większy odsetek rezygnujacych klientów, ale jest to również naturalne, ponieważ ten rodzaj umowy pozwala na szybką rezygnację lub rezygnację w dowolnym momencie. Nie daje to natomiast odpowiedzi, jaki faktycznie był okres jej trwania

**Staż a churn**
Czy odchodzą nowi czy starzy klienci?

```python
df["tenure_group"] = pd.cut(
    df["tenure"],
    bins=[-1, 6, 12, 24, 48, 72],
    labels=["0-6", "7-12", "13-24", "25-48", "49-72"],
)

fig, ax = plt.subplots(figsize=(8, 5))

sns.barplot(
    data=df, 
    x="tenure_group", 
    y="Churn", 
    errorbar=None, 
    ax=ax
)

ax.set_title("Odsetek odejść według stażu")
ax.set_xlabel("Staż [miesiące]")
ax.set_ylabel("Odsetek klientów, którzy odeszli")

plt.tight_layout()
plt.show()
```

Najwięcej odejść jest w pierszych 6 miesiącach od podpisania umowy, potem sukcesywnie się to zmiejsza.

**Opłata miesięczna a churn**

Czy odchodzący płacą inaczej? Jak rozkładają się wszyscy odchodzący (1869 osób = 100%)

```python
df["charges_group"] = pd.cut(
    df["MonthlyCharges"],
    bins=[0, 30, 50, 70, 90, 120],
    labels=["do 30", "30-50", "50-70", "70-90", "powyżej 90"],
)

churned = df[df["Churn"] == 1]

share = (
    churned["charges_group"]
    .value_counts(normalize=True)
    .sort_index()
    .reset_index()
)
share.columns = ["charges_group", "share"]
counts = churned["charges_group"].value_counts().sort_index().values

fig, ax = plt.subplots(figsize=(8, 5))

sns.barplot(
    data=share, 
    x="charges_group", 
    y="share", 
    color="steelblue", 
    ax=ax
)

for i, (s, n) in enumerate(zip(share["share"], counts)):
    ax.text(i, s + 0.01, f"{s:.0%}\n(n={n})", ha="center")

ax.set_title("Z którego przedziału opłat odeszło najwięcej klientów")
ax.set_xlabel("Opłata miesięczna [$]")
ax.set_ylabel("Udział w ogólnej liczbie odejść")
ax.set_ylim(0, share["share"].max() + 0.1)

plt.tight_layout()
plt.show()
```

Najwięcej odeszło z grup 70-90 oraz powyżej 90, odpowiednio 697 i 570 osób ze wszystkich, który odeszli.

**Chrum a umowa a opłata miesięczna**

```python
charges_contract_stats = (
    df.groupby(["charges_group", "Contract"], observed=True)["Churn"]
    .agg(churn_rate="mean", n="size")
    .reset_index()
)

fig, ax = plt.subplots(figsize=(12, 6))

sns.barplot(
    data=charges_contract_stats,
    x="charges_group",
    y="churn_rate",
    hue="Contract",
    hue_order=["Month-to-month", "One year", "Two year"],
    palette="Blues_d",
    ax=ax,
)

for p in ax.patches:
    height = p.get_height()
    if not pd.isna(height) and height > 0:
        ax.annotate(
            f"{height:.1%}",  
            (p.get_x() + p.get_width() / 2.0, height),
            ha="center",
            va="bottom",
            fontsize=8,
            xytext=(0, 2),
            textcoords="offset points",
        )

ax.set_title("Odsetek odejść według opłaty miesięcznej i typu umowy")
ax.set_xlabel("Opłata miesięczna [$]")
ax.set_ylabel("Odsetek klientów, którzy odeszli")
ax.set_ylim(0, charges_contract_stats["churn_rate"].max() + 0.1)
ax.legend(title="Typ umowy")

plt.tight_layout()
plt.show()
```

- *duży odsetek churnu w przedziale 30-50* - umowy krótkoterminowe (jak w każdym przedziale) stanowią najwyższy czynnik ryzyka. Jednocześnie klienci na umowach rocznych (8,9%) i dwuletnich (2,7%) w tej grupie praktycznie nie odchodzą, ale jest ich na tyle mało, że wysoki odsetek Month-to-month podciąga średnią całej grupy w górę. Ergo - klienci nie są przywiązani umowami, a kwota rzędu 30-50$ to dla nich moment, w którym łatwo rezygnują, jeśli oferta przestaje im odpowiadać.
  
- *przedziały 70-90 oraz powyżej 90* - w obu przypadkach dla umów krótkoterminowych mamy odsetek churmu powyżej 50%, więc ponad połowa klientów z takimi umowami znajduje się w grupie klientów płacącyc najwyższe stawki, ale nie została przywiazana umową długoterminową. To grupa generująca najwyższe straty dla firmy, o najwyższym ryzyku chrumu.

**Obciążenia a churn**

Jak sytuacja życiowa klienta (senior, partner, dzieci/osoby na utrzymaniu) oraz koszty i staż wpływają na odejście? Dwa panele:

*Lewy*: odsetek odejść dla profilu gospodarstwa (kombinacja SeniorCitizen, Partner, Dependents). W etykiecie jest liczebność grupy.


*Prawy*: mapa ciepła odsetka odejść w zależności od stażu i opłaty miesięcznej.

```python
burden = df[["SeniorCitizen", "Partner", "Dependents",
             "MonthlyCharges", "tenure", "Churn"]].copy()


burden["profile"] = (
    burden["SeniorCitizen"].map({0: "", 1: "Senior, "})
    + "Partner: " + burden["Partner"]
    + ", Dzieci: " + burden["Dependents"]
)

profile_stats = (
    burden.groupby("profile")["Churn"]
    .agg(churn_rate="mean", n="size")
    .sort_values("churn_rate", ascending=False)
    .reset_index()
)
profile_stats["label"] = profile_stats["profile"] + " (n=" + profile_stats["n"].astype(str) + ")"


burden["tenure_bin"] = pd.cut(
    burden["tenure"], bins=[-1, 12, 24, 48, 72],
    labels=["0-12", "13-24", "25-48", "49-72"],
)
burden["charges_bin"] = pd.qcut(
    burden["MonthlyCharges"], q=4,
    labels=["najniższe", "niskie", "wysokie", "najwyższe"],
)

heat = burden.pivot_table(
    index="tenure_bin", columns="charges_bin",
    values="Churn", aggfunc="mean", observed=True,
)

fig, axes = plt.subplots(1, 2, figsize=(16, 6))

sns.barplot(data=profile_stats, y="label", x="churn_rate", color="steelblue", ax=axes[0])
axes[0].set_title("Odsetek odejść według profilu gospodarstwa")
axes[0].set_xlabel("Odsetek klientów, którzy odeszli")
axes[0].set_ylabel("")

sns.heatmap(heat, annot=True, fmt=".2f", cmap="Reds", vmin=0, vmax=1, ax=axes[1])
axes[1].set_title("Odsetek odejść: staż a opłata miesięczna")
axes[1].set_xlabel("Opłata miesięczna (kwartyle)")
axes[1].set_ylabel("Staż [miesiące]")

plt.tight_layout()
plt.show()
```

- *Odsetek odejść według profilu gospodarstwa* - Najwyższy odsetek odejść notują zazwyczaj osoby samotne, młodsze (niebędące seniorami), bez osób na utrzymaniu, podczas gdy klienci mieszkający z rodzinami/partnerami i dziećmi są znacznie bardziej stabilni (wykazują niższy churn). Natomiast należy pamiętać, że w danych nie ma informacji ile churnu dotyczącego seniorów związane jest ze zgonami (tu churn sięga 50% wśród samotnych seniorów).

```python

addons = ["OnlineSecurity", "OnlineBackup", "DeviceProtection",
          "TechSupport", "StreamingTV", "StreamingMovies"]

df["n_products"] = (
    df["PhoneService"].eq("Yes").astype(int)
    + df["InternetService"].ne("No").astype(int)
    + df["MultipleLines"].eq("Yes").astype(int)
    + sum(df[c].eq("Yes").astype(int) for c in addons)
)

by_count = (
    df.groupby("n_products")["Churn"]
    .agg(churn_rate="mean", n="size")
    .reset_index()
)

fig, ax = plt.subplots(figsize=(9, 5))

sns.barplot(data=by_count, x="n_products", y="churn_rate",
            color="steelblue", ax=ax)
for i, row in by_count.iterrows():
    ax.text(i, row["churn_rate"] + 0.01, f"n={row['n']}", ha="center", fontsize=8)

ax.set_title("Odsetek odejść według liczby produktów")
ax.set_xlabel("Liczba produktów klienta")
ax.set_ylabel("Odsetek klientów, którzy odeszli")
ax.set_ylim(0, by_count["churn_rate"].max() + 0.1)

plt.tight_layout()
plt.show()
```

```python
pay_stats = (
    df.groupby("PaymentMethod")["Churn"]
    .agg(churn_rate="mean", n="size")
    .sort_values("churn_rate", ascending=False)
    .reset_index()
)

fig, ax = plt.subplots(figsize=(9, 5))

sns.barplot(data=pay_stats, x="PaymentMethod", y="churn_rate",
            color="steelblue", ax=ax)
for i, row in pay_stats.iterrows():
    ax.text(i, row["churn_rate"] + 0.01, f"n={row['n']}", ha="center", fontsize=8)

ax.set_title("Odsetek odejść według metody płatności")
ax.set_xlabel("Metoda płatności")
ax.set_ylabel("Odsetek klientów, którzy odeszli")
ax.set_ylim(0, pay_stats["churn_rate"].max() + 0.1)
ax.tick_params(axis="x", rotation=15)

plt.tight_layout()
plt.show()
```

```python
# usługa internetowa
net_stats = (
    df.groupby("InternetService")["Churn"]
    .agg(churn_rate="mean", n="size")
    .sort_values("churn_rate", ascending=False)
    .reset_index()
)

# DSL vs Fiber według liczby produktów (grupy n >= 30) 
by_internet = (
    df[df["InternetService"] != "No"]
    .groupby(["InternetService", "n_products"])["Churn"]
    .agg(churn_rate="mean", n="size")
    .reset_index()
)
by_internet = by_internet[by_internet["n"] >= 30]

# każdy produkt osobno (ma / nie ma)
all_rows = pd.Series(True, index=df.index)
has_phone = df["PhoneService"] == "Yes"
has_net = df["InternetService"] != "No"

specs = [
    ("PhoneService", df["PhoneService"].eq("Yes"), all_rows),
    ("MultipleLines", df["MultipleLines"].eq("Yes"), has_phone),
    ("Internet DSL", df["InternetService"].eq("DSL"), all_rows),
    ("Internet Fiber", df["InternetService"].eq("Fiber optic"), all_rows),
] + [(c, df[c].eq("Yes"), has_net) for c in addons]

rows = []
for name, has, pop in specs:
    rows.append({"produkt": name, "grupa": "ma",
                 "churn_rate": df.loc[pop & has, "Churn"].mean(),
                 "n": (pop & has).sum()})
    rows.append({"produkt": name, "grupa": "nie ma",
                 "churn_rate": df.loc[pop & ~has, "Churn"].mean(),
                 "n": (pop & ~has).sum()})
product_effect = pd.DataFrame(rows)

fig, axes = plt.subplots(3, 1, figsize=(10, 18))

sns.barplot(data=net_stats, x="InternetService", y="churn_rate",
            color="steelblue", ax=axes[0])
for i, row in net_stats.iterrows():
    axes[0].text(i, row["churn_rate"] + 0.01, f"n={row['n']}", ha="center", fontsize=9)
axes[0].set_title("Odejścia według usługi internetowej")
axes[0].set_xlabel("Usługa internetowa")
axes[0].set_ylabel("Odsetek klientów, którzy odeszli")
axes[0].set_ylim(0, net_stats["churn_rate"].max() + 0.1)

sns.lineplot(data=by_internet, x="n_products", y="churn_rate",
             hue="InternetService", marker="o", ax=axes[1])
axes[1].set_title("Ilość produktów a odejście klienta, osobno dla DSL i Fiber optic")
axes[1].set_xlabel("Liczba produktów klienta")
axes[1].set_ylabel("Odsetek klientów, którzy odeszli")
axes[1].set_ylim(0, 1)

sns.barplot(data=product_effect, y="produkt", x="churn_rate", hue="grupa",
            palette={"ma": "steelblue", "nie ma": "lightgray"}, ax=axes[2])
axes[2].set_title("Odsetek odejść: klienci z produktem i bez niego")
axes[2].set_xlabel("Odsetek klientów, którzy odeszli")
axes[2].set_ylabel("")

plt.tight_layout()
plt.show()

product_effect.pivot(index="produkt", columns="grupa", values=["churn_rate", "n"]).round(3)
```
